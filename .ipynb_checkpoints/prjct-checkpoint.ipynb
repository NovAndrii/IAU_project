V každom súbore sa líšia oddeľovače, preto pre každý súbor zvlášť vytvárame dátový rámec

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path


pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

sns.set_theme(style="whitegrid")

RANDOM_STATE = 42

data_path = Path("Datasets/037")

files = [
    file
    for file in data_path.rglob("*")
    if file.is_file()
]

print("Number of files:", len(files))

for file in files:
    print(file)

gateways = pd.read_csv(
    data_path / "gateways.csv",
    sep=';',
)

readings = pd.read_csv(
    data_path / "readings.csv",
    sep='\t',
)

units = pd.read_csv(
    data_path / "units.csv",
    sep=',',
)


datasets = {
    "gateways": gateways,
    "readings": readings,
    "units": units
}

for name, data in datasets.items():
    print(f"\n{name.upper()}")
    print("Shape:", data.shape)
    print("Columns:", data.columns.tolist())
    display(data.head())


print("Gateways:", gateways.shape)
print("Units:", units.shape)
print("Readings:", readings.shape)
```

Vidíme problém – dátumy v tabuľke sú uložené vo formáte str, a nie datetime

```python
x = pd.crosstab(gateways['Technician'], gateways['id_depot'])
print(x)
```

Vzhľadom na malý počet jedinečných prvkov usudzujeme, že ide o "Categorical data(Nominal data)".

V data sete "gateways" je hneď zrejmé, že formát v stĺpcoch „seen_last“ a „date_install“ sa v jednotlivých riadkoch líši, preto ich zjednotíme do jedného formátu

```python
gateways["seen_last"] = pd.to_datetime(gateways["seen_last"],format = "mixed", errors="coerce")
gateways["date_install"] = pd.to_datetime(gateways["date_install"],format = "mixed", errors="coerce")
```

```python
num = gateways.select_dtypes("number").drop(columns=["id_unit", "id_depot"])
corr = num.corr(method="spearman")
print(corr)
```

```python
sns.clustermap(corr, cmap="vlag", center = 0, annot = len(corr) < 15)
```

```python
pairs = (corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
             .stack().dropna()
             .sort_values(key=abs, ascending=False))
```

```python
missing_gateways = pd.DataFrame({
    "missing_count": gateways.isna().sum(),
    "missing_percentage": (
        gateways.isna().mean() * 100
    ).round(2)
})



plt.figure(figsize=(12, 5))

sns.barplot(
    x=missing_gateways.index,
    y=missing_gateways["missing_percentage"]
)

plt.xticks(rotation=45)
plt.ylabel("Missing values (%)")
plt.title("Missing values in gateways")
plt.tight_layout()
plt.show()
```
