Na boxplote sú viditeľné outliery, preto ich odstránime

````python
def identify_outliers(a):
    lower = a.quantile(0.25) - 1.5 * stats.iqr(a)
    upper = a.quantile(0.75) + 1.5 * stats.iqr(a)

    return a[(a > upper) | (a < lower)]
````

````python
test_data = gateways['dbm_signal']
dbm_signal_out = identify_outliers(test_data)
dbm_signal_out.info()
````

````python
sns.boxplot(data = dbm_signal_out)
````

````python
readings_eda = readings.copy()

readings_eda["oilTemperature"] = readings_eda[
    "oilTemperature"
].replace(-999, np.nan)

readings_eda["motorCurrent"] = readings_eda["motorCurrent"].replace(99, np.nan)

readings_eda.hist(
    figsize=(15, 12),
    bins=30
)

plt.tight_layout()
plt.show()
````

#### Po dočasnom odstránení podozrivých hodnôt oilTemperature = -999 a motorCurrent = 99 boli grafy prehľadnejšie.
````python
fig, axes = plt.subplots(4, 3, figsize=(15, 15))
axes = axes.flatten()

for i, column in enumerate(sensors):
    sns.boxplot(
        data=readings_eda,
        x=column,
        ax=axes[i]
    )
    axes[i].set_title(column)

plt.tight_layout()
plt.show()
````

## Vychýlené hodnoty

````python
SENSORS = ["motorCurrent", "filter_dp", "oilTemperature", "pressure_reservoir",
           "pressure_working", "pressureOil", "vibrationRms", "flowmeter",
           "intake_humidity", "ambientNoise", "bearingTemp", "hoursService", "dbm_signal"]

def iqr_bounds(s, k=1.5):
    q1, q3 = s.quantile([0.25, 0.75])
    return q1 - k * (q3 - q1), q3 + k * (q3 - q1)

rows = []
for c in SENSORS:
    lo, hi = iqr_bounds(df[c])
    m = (df[c] < lo) | (df[c] > hi)
    rows.append({"atribút": c, "dolná": round(lo, 2), "horná": round(hi, 2),
                 "počet": m.sum(), "%": round(100 * m.mean(), 2),
                 "failure vo vychýlených": round(df.loc[m, "failure"].mean(), 3) if m.any() else np.nan,
                 "failure v ostatných": round(df.loc[~m, "failure"].mean(), 3)})
outl = pd.DataFrame(rows).set_index("atribút")
display(outl)
````

Technika 1: ODSTRÁNENIE riadkov s vychýlenou hodnotou (IQR 1.5)

````python
any_out = pd.Series(False, index=df.index)
for c in SENSORS:
    lo, hi = iqr_bounds(df[c])
    any_out |= (df[c] < lo) | (df[c] > hi)
df_removed = df[~any_out]
````

Technika 2: NAHRADENIE hraničnými hodnotami rozdelenia (5 % a 95 % kvantil) – winsorizácia

````python
df_wins = df.copy()
for c in SENSORS:
    lo, hi = df[c].quantile([0.05, 0.95])
    df_wins[c] = df[c].clip(lo, hi)
````

Tabuľka porovnáva vplyv oboch techník na počet záznamov a na cieľovú premennú.

````python
cmp = pd.DataFrame({
    "deleted rows": [len(df), len(df_removed), len(df_wins)],
    "number of failure=1": [df.failure.sum(), df_removed.failure.sum(), df_wins.failure.sum()],
    "disrt failure": [df.failure.mean(), df_removed.failure.mean(), df_wins.failure.mean()],
}, index=["pôvodné", "T1: odstránenie (IQR)", "T2: winsorizácia 5/95"]).round(4)
display(cmp)
````

Tabuľka porovnáva bodovo-biseriálnu koreláciu (Pearsonov koeficient s binárnou premennou) každého senzora s failure pred a po úprave vychýlených hodnôt.

````python
eff = pd.DataFrame({
    name: [d[c].corr(d["failure"]) for c in SENSORS]
    for name, d in [("pôvodné", df), ("T1 odstránenie", df_removed), ("T2 winsorizácia", df_wins)]
}, index=SENSORS).round(3)
display(eff.sort_values("pôvodné", key=abs, ascending=False))

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
for ax, (name, d) in zip(axes, [("pôvodné", df), ("T1 odstránenie", df_removed), ("T2 winsorizácia 5/95", df_wins)]):
    sns.boxplot(data=d, x="failure", y="filter_dp", ax=ax)
    ax.set_title(f"filter_dp – {name}")
plt.tight_layout(); plt.show()
````

#### **Porovnanie techník a rozhodnutie**

- Pravidlo IQR označuje najviac hodnôt v filter_dp a vibrationRms, čo zodpovedá ich pravostrannému zošikmeniu. Pri filter_dp je medzi vychýlenými hodnotami *podiel porúch ~25 %, teda dvojnásobok* oproti ostatným meraniam (~12 %).
- *T2 – winsorizácia 5/95* zachová všetky riadky aj všetky poruchy. Orezaním 10 % hodnôt na kvantily však sploští chvost rozdelenia, kde sú poruchy sústredené.
- *Rozhodnutie:* vychýlené hodnoty v senzoroch nie sú chyby merania. Chybové kódy (−999, 99) sme už odstránili a ostatné hodnoty ležia vo fyzikálne prípustnom rozsahu. Ide o skutočné extrémne stavy zariadenia, ktoré *signalizujú blížiacu sa poruchu*. Preto ich v dátach *ponechávame*.

Pre každý kategorický atribút sme χ² testom nezávislosti overili, či súvisí s poruchou (H₀: atribút a failure sú nezávislé; H₁: sú závislé; α = 0.05). Silu závislosti vyjadruje Cramérovo V (0 až 1).

````python
def cramers_v(x, y):
    ct = pd.crosstab(x, y)
    chi2, p, dof, _ = stats.chi2_contingency(ct)
    return np.sqrt(chi2 / (ct.values.sum() * (min(ct.shape) - 1))), p

CAT = ["class_duty", "lineCode", "Manufacturer", "modelHardware", "versionFirmware",
         "filter_dp_missing", "vibrationRms_missing"]
rows = []
for c in CAT:
    v, p = cramers_v(df[c], df["failure"])
    rates = df.groupby(c)["failure"].mean()
    rows.append({"atribút": c, "kategórií": df[c].nunique(), "Cramérovo V": v, "p (χ²)": p,
                 "min podiel failure": rates.min(), "max podiel failure": rates.max()})
cat_vs_target = pd.DataFrame(rows).set_index("atribút").sort_values("Cramérovo V", ascending=False)
display(cat_vs_target.style.format({"Cramérovo V": "{:.3f}", "p (χ²)": "{:.1e}",
                                    "min podiel failure": "{:.3f}", "max podiel failure": "{:.3f}"}))

fig, axes = plt.subplots(1, 4, figsize=(15, 4))
for ax, c in zip(axes, ["class_duty", "lineCode", "modelHardware", "versionFirmware"]):
    df.groupby(c)["failure"].mean().plot.bar(ax=ax, rot=0)
    ax.axhline(df.failure.mean(), ls="--", color="grey"); ax.set_title(f"podiel failure podľa {c}")
plt.tight_layout(); plt.show()
````

Atribút považujeme za súvisiaci s poruchou, ak oba grafy ukazujú rovnaký smer.

  - hoursService: na bodovom grafe sú poruchy rozložené rovnomerne až do 45 000 h, bez

*Záver:* oba typy grafov sa zhodujú. Najsilnejší vzťah k poruche majú hoursService,
filter_dp, motorCurrent_rel, bearingTemp a tlaky. motorCurrent treba posudzovať

# štatistické overenie hypotéz

### Priemerná hodnota bearingTemp sa líši medzi poruchovým stavom a normálnou prevádzkou.
#### H0:Priemerná hodnota bearingTemp počas poruchy sa nelíši od priemernej hodnoty počas normálnej prevádzky.
#### H1: Priemerná hodnota bearingTemp počas poruchy sa líši od priemernej hodnoty počas normálnej prevádzky.


````python
data = readings.copy()
display(data[["failure", "bearingTemp"]].head())

print("Failure values:", data["failure"].unique())
print("Missing failure values:", data["failure"].isna().sum())
print("Missing bearingTemp values:", data["bearingTemp"].isna().sum())
````

````python
bearing_normal = data.loc[
    data["failure"] == 0,
    "bearingTemp"
].dropna()

bearing_failure = data.loc[
    data["failure"] == 1,
    "bearingTemp"
].dropna()

print("Normal observations:", len(bearing_normal))
print("Failure observations:", len(bearing_failure))
````

````python
bearing_summary = data.groupby("failure")["bearingTemp"]
bearing_summary.describe()
````

##### Priemerná hodnota bearingTemp bola 64,81 počas normálnej prevádzky a 66,61 počas poruchy. Medián bol počas poruchy tiež vyšší. Výsledky naznačujú, že bearingTemp môže byť počas poruchy vyššia, ale na potvrdenie je potrebný štatistický test.

````python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x="failure",
    y="bearingTemp"
)

plt.title("Bearing temperature by failure state")
plt.xlabel("Failure")
plt.ylabel("Bearing temperature")
plt.show()
````

````python


fig, axes = plt.subplots(1, 2, figsize=(12, 5))

stats.probplot(bearing_normal, dist="norm", plot=axes[0])
axes[0].set_title("Normal operation: Q-Q plot")

stats.probplot(bearing_failure, dist="norm", plot=axes[1])
axes[1].set_title("Failure: Q-Q plot")

plt.tight_layout()
plt.show()
````

#### Boxplot ukazuje, že bearingTemp je počas poruchy mierne vyššia. Q–Q ploty ukazujú, že obe skupiny majú približne normálne rozdelenie, iba s malými odchýlkami na okrajoch. Preto je možné použiť t-test. Q–Q plot v tomto prípade poskytuje dostatočne spoľahlivé výsledky

````python
stat_bearing, p_bearing = stats.shapiro(bearing_normal)
print(f"bearingTemp: p-value = {p_bearing:.4g}")

````


#### Shapiro-Wilkov test dosiahol p-hodnotu približne 0,3 (> 0,05), preto nezamietame H0 o normalite rozdelenia. Údaje teda nevykazujú štatisticky významnú odchýlku od normálneho rozdelenia. Pri veľkom počte pozorovaní však môže byť test príliš citlivý aj na malé odchýlky, preto výsledok posudzujeme aj pomocou Q-Q grafu.


````python
levene_result = stats.levene(
    bearing_normal,
    bearing_failure
)

print("Levene statistic:", levene_result.statistic)
print("p-value:", levene_result.pvalue)
````

#### p >= 0,05 → medzi rozptylmi skupín nie je štatisticky významný rozdiel.

````python
bearing_test = stats.ttest_ind(
    bearing_failure,
    bearing_normal,
    equal_var=True,
    alternative="greater"
)

print("Test statistic:", bearing_test.statistic)
print("p-value:", bearing_test.pvalue)
````

#### T-test dosiahol p-value nižšiu ako 0,05, preto bola nulová hypotéza zamietnutá. Existuje štatisticky významný dôkaz, že priemerná hodnota bearingTemp je počas poruchy odlišna.

````python


n_failure = len(bearing_failure)
n_normal = len(bearing_normal)

pooled_std = np.sqrt(
    (
        (n_failure - 1) * bearing_failure.var(ddof=1)
        + (n_normal - 1) * bearing_normal.var(ddof=1)
    )
    / (n_failure + n_normal - 2)
)

cohens_d = (
    bearing_failure.mean() - bearing_normal.mean()
) / pooled_std

print("Cohen's d:", cohens_d)
````

#### Cohen’s d je 0,226, čo predstavuje malý effect size. Hodnota bearingTemp je teda počas poruchy odlišna, ale rozdiel medzi skupinami je relatívne malý.

````python
stat_bearing, p_bearing = stats.ttest_ind(
    bearing_failure,
    bearing_normal,
    equal_var=True,
    alternative="greater"
)

print(
    "Ttest_indResult: statistic=",
    stat_bearing,
    "pvalue=",
    p_bearing
)

power_bearing = TTestIndPower().power(
    effect_size=abs(cohens_d),
    nobs1=len(bearing_failure),
    alpha=0.05,
    ratio=len(bearing_normal) / len(bearing_failure),
    alternative="larger"
)

print("Bearing temperature power:", power_bearing)
````

#### Statistical power testu pre bearingTemp bola približne 1,00. Keďže je táto hodnota vyššia ako 0,80, test má dostatočnú statistical power.


#### Shapiro-Wilkov test dosiahol p-hodnotu približne 0,6 (> 0,05), preto nezamietame H0 o normalite rozdelenia. Údaje teda nevykazujú štatisticky významnú odchýlku od normálneho rozdelenia. Pri veľkom počte pozorovaní však môže byť test príliš citlivý aj na malé odchýlky, preto výsledok posudzujeme aj pomocou Q-Q grafu.


````python
levene_pressure = stats.levene(
    pressure_normal,
    pressure_failure
)

print("Levene statistic:", levene_pressure.statistic)
print("p-value:", levene_pressure.pvalue)
````

#### variances oboch skupín sú rovnaké.

````python
pressure_test = stats.ttest_ind(
    pressure_failure,
    pressure_normal,
    equal_var=True,
    alternative="less"
)

print("Test statistic:", pressure_test.statistic)
print("p-value:", pressure_test.pvalue)
````

#### T-test dosiahol test statistic −8,77 a p-value nižšiu ako 0,05. Nulová hypotéza bola preto zamietnutá. Výsledky ukazujú, že priemerná hodnota pressureOil je počas poruchy štatisticky odlišna.

````python
n_failure = len(pressure_failure)
n_normal = len(pressure_normal)

pooled_std = np.sqrt(
    (
        (n_failure - 1) * pressure_failure.var(ddof=1)
        + (n_normal - 1) * pressure_normal.var(ddof=1)
    )
    / (n_failure + n_normal - 2)
)

cohens_d_pressure = (
    pressure_failure.mean() - pressure_normal.mean()
) / pooled_std

print("Cohen's d:", abs(cohens_d_pressure))
````

### Cohen’s d je −0,239, čo predstavuje malý effect size. Záporná hodnota ukazuje, že pressureOil je počas poruchy odlišní. Rozdiel je teda štatisticky významný, ale jeho veľkosť je relatívne malá.

````python
stat_bearing, p_bearing = stats.ttest_ind(
    bearing_failure,
    bearing_normal,
    equal_var=True,
    alternative="greater"
)

print(
    "Ttest_indResult: statistic=",
    stat_bearing,
    "pvalue=",
    p_bearing
)

power_pressure = TTestIndPower().power(
    effect_size=abs(cohens_d_pressure),
    nobs1=len(pressure_failure),
    alpha=0.05,
    ratio=len(pressure_normal) / len(pressure_failure),
    alternative="larger"
)

print("Oil pressure power:", power_pressure)
````
