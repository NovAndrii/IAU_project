Na boxplote sú viditeľné outliery, preto ich odstránime

```python
def identify_outliers(a):
    lower = a.quantile(0.25) - 1.5 * stats.iqr(a)
    upper = a.quantile(0.75) + 1.5 * stats.iqr(a)

    return a[(a > upper) | (a < lower)]
```

```python
test_data = gateways['dbm_signal']
dbm_signal_out = identify_outliers(test_data)
dbm_signal_out.info()
```

```python
sns.boxplot(data = dbm_signal_out)
```

#### The statistics reveal potentially abnormal sensor values, particularly oilTemperature = -999 and motorCurrent = 99, which require further investigation during data cleaning.

```python
sensors = [
    "motorCurrent",
    "filter_dp",
    "oilTemperature",
    "pressure_reservoir",
    "pressure_working",
    "pressureOil",
    "vibrationRms",
    "flowmeter",
    "intake_humidity",
    "ambientNoise",
    "bearingTemp"
]


readings[sensors].hist(
    figsize=(15, 12),
    bins=30
)

plt.tight_layout()
plt.show()

```

#### The sensor attributes have different distribution shapes. pressure_reservoir, pressure_working, pressureOil, flowmeter, ambientNoise, and bearingTemp show approximately normal distributions. filter_dp and vibrationRms are positively skewed. . The motorCurrent distribution is distorted by the extreme value 99 while oilTemperature is distorted by the invalid value −999.

```python
readings_eda = readings.copy()

readings_eda["oilTemperature"] = readings_eda[
    "oilTemperature"
].replace(-999, np.nan)

readings_eda["motorCurrent"] = readings_eda["motorCurrent"].replace(99, np.nan)

readings_eda.hist(
    figsize=(15, 12),
    bins=30
)

plt.tight_layout()
plt.show()
```

#### After temporarily removing the suspicious values oilTemperature = -999 and motorCurrent = 99, the graphs became clearer. Most oil temperatures are around 55, so −999 is probably an invalid value. Most motor current values are between 1 and 10, so 99 is probably abnormal

```python
rd = readings.copy()


Q1 = rd['filter_dp'].quantile(0.25)
Q3 = rd['filter_dp'].quantile(0.75)
IQR = Q3 - Q1


lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

rd.loc[
    (rd["filter_dp"] < lower_bound) |
    (rd["filter_dp"] > upper_bound),
    "filter_dp"
] = float("nan")

Q1 = rd['vibrationRms'].quantile(0.25)
Q3 = rd['vibrationRms'].quantile(0.75)
IQR = Q3 - Q1


lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

rd.loc[
    (rd["vibrationRms"] < lower_bound) |
    (rd["vibrationRms"] > upper_bound),
    "vibrationRms"
] = float("nan")


rd['id_unit'] = rd['id_unit'].str.extract(r'(\d+)')[0].astype(int)
rd.loc[rd['id_unit'] > 80000, 'id_unit'] = np.nan

rd.loc[rd['motorCurrent'] >= 99] = np.nan
rd.loc[rd['oilTemperature'] <= -999] = np.nan
rd['ts_cycle'] = pd.to_datetime(rd['ts_cycle'], format="mixed", errors="coerce")

rd.hist(
    figsize=(15, 12),
    bins=30
)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 3, figsize=(15, 15))
axes = axes.flatten()

for i, column in enumerate(sensors):
    sns.boxplot(
        data=readings_eda,
        x=column,
        ax=axes[i]
    )
    axes[i].set_title(column)

plt.tight_layout()
plt.show()
```

#### The boxplots show many possible outliers, especially in filter_dp and vibrationRms. Some possible outliers are also present in the temperature, pressure, flow, and noise attributes. However, these values are not necessarily incorrect and will be checked later during data cleaning.
